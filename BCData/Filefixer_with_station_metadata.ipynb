```python
maindata = pd.DataFrame()
for paths in subpath[1:]:
    files = os.listdir(paths)
    for f in files:
        data = pd.read_csv(paths + '/' + f, skiprows=1)
        data['file'] = paths + '/' + f
        maindata = pd.concat([maindata, data])
```

```python
metadata = pd.read_csv('station-metadata-by-history (1).csv')
metadata
```

```python
newtable = pd.DataFrame()
iter = 1
for index, row in maindata.iterrows():
    if (iter % 100 == 0):
        print(iter)
        # break
    splits = row['file'].split('/')
    group = splits[-2]
    areaid = splits[-1].split('.')[0]
    data = metadata.loc[metadata['Network Name'] == group].loc[metadata['Native ID'] == areaid]
    # print(group, areaid, row['file'])
    # print(data)
    try:
        row['Station Name'] = data['Station Name'].iloc[0]
        row['Longitude'] = data['Longitude'].iloc[0]
        row['Latitude'] = data['Latitude'].iloc[0]
        newtable = pd.concat([newtable, row], axis=1)
    except:
        print('Skipping ' + str(group) + ' ' + str(areaid))
    iter += 1
```
