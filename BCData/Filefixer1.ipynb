```python
import pandas as pd
data = pd.read_csv('EMSWR_Rus_044403.csv')
data.head()
```

```python
locations = data['LOCATION NAME'].unique()
len(locations)
```

```python
dataprocessed = pd.DataFrame()
i = 1
for location in locations:
    print(i, location)
    i += 1
    subset = data.loc[data['LOCATION NAME'] == location]
    tempdf = pd.DataFrame()
    tempdf['LOCATION'] = [location]
    tempdf['LONGITUDE'] = [-subset['LONGITUDE (Decimal Degrees)'].iloc[0]]
    tempdf['LATITUDE'] = [subset['LATITUTE (Decimal Degrees)'].iloc[0]]
    tempdf['Time'] = [subset['START DATE']]
    
    for param in subset['PARAMETER NAME']:
        if pd.isna(param):
            continue
        tempdf[str(param)] = [subset.loc[subset['PARAMETER NAME'] == param, 'Results'].iloc[0]]
    dataprocessed = pd.concat([dataprocessed, tempdf], ignore_index=True)
```

```python
dataprocessed.to_excel('Processed_data.xlsx')
```
