```python
import numpy as np
import pandas as pd
processed = pd.read_excel('Processed_dataNew.xlsx.xlsx')
fused = pd.read_excel('Fused_data.xlsx')
fused = fused.transpose()
data = fused.copy()
data.columns = data.iloc[0] 
data = data[1:]
data = data.reset_index(drop=True)
processed['Precip_Climatology'] = np.nan
processed['T_mean_Climatology'] = np.nan
processed['Tn_Climatology'] = np.nan
processed['Tx_Climatology'] = np.nan
```

```python
for i in range(len(processed)):
    subset = processed.iloc[i, :]
    location = subset.LOCATION
    x = subset['LONGITUDE']
    y = subset['LATITUDE']
    point1 = np.asarray([x, y])
    print(location, x, y)
    closest_distance = np.inf
    bestj = -1
    for j in range(len(data)):
        subset2 = data.iloc[j, :]
        data_x = subset2['Longitude']
        data_y = subset2['Latitude']
        point2 = np.asarray([data_x, data_y])
        dist = np.linalg.norm(point1 - point2)
        if dist < closest_distance:
            closest_distance = dist
            bestj = j

    bestval = data.iloc[bestj, :]
    processed.loc[processed['LOCATION']==location, 'Precip_Climatology'] = bestval['Precip_Climatology']
    processed.loc[processed['LOCATION']==location, 'T_mean_Climatology'] = bestval[' T_mean_Climatology']
    processed.loc[processed['LOCATION']==location, 'Tn_Climatology'] = bestval[' Tn_Climatology']
    processed.loc[processed['LOCATION']==location, 'Tx_Climatology'] = bestval[' Tx_Climatology']
    processed.loc[processed['LOCATION']==location, 'Nearest_Longitude'] = bestval['Longitude']
    processed.loc[processed['LOCATION']==location, 'Nearest_Latitude'] = bestval['Latitude']

    # break
    
processed
```
