```python
dataprocessed = pd.DataFrame()
i = 1
data = data1
locations = data['LOCATION NAME'].unique()
for location in locations:
    print(i, location)
    i += 1
    subset = data.loc[data['LOCATION NAME'] == location]
    times = subset['START DATE'].unique()
    for time in times:
        timeslice = subset.loc[subset['START DATE'] == time, ['Location', 'OBSWell', 'LONGITUDE (Decimal Degrees)', 'LATITUTE (Decimal Degrees)', 'Date', 'PARAMETER NAME', 'Results']].copy()
        tempdf = pd.DataFrame()
        tempdf['LOCATION'] = [timeslice['Location'].iloc[0]]
        tempdf['OBSWELL'] = [timeslice['OBSWell'].iloc[0]]
        tempdf['LONGITUDE'] = [timeslice['LONGITUDE (Decimal Degrees)'].iloc[0]]
        tempdf['LATITUTE'] = [timeslice['LATITUTE (Decimal Degrees)'].iloc[0]]
        tempdf['DATE'] = [timeslice['Date'].iloc[0]]

        for param in timeslice['PARAMETER NAME']:
            if pd.isna(param):
                continue
            tempdf[str(param)] = [timeslice.loc[timeslice['PARAMETER NAME'] == param, 'Results'].iloc[0]]
        dataprocessed = pd.concat([dataprocessed, tempdf], ignore_index=True)

# dataprocessed
```

```python
dataprocessednew = dataprocessed.copy()
cols_to_delete4 = ["p;p'-DDD", "p;p'-DDE", "p;p'-DDT", "Sulfide Total","Solids Total","Wtr Lev"]
dataprocessed = dataprocessed.drop(cols_to_delete4, axis=1, errors='ignore')
dataprocessed = dataprocessed[~((dataprocessed['Fe-D'].isna()) & (dataprocessed['Pb-T'].isna()) & (dataprocessed['Co-D'].isna()))]

dataprocessed['U--T'].fillna(dataprocessed['U--D'], inplace=True)
dataprocessed['As-T'].fillna(dataprocessed['As-D'], inplace=True)
dataprocessed['Cd-T'].fillna(dataprocessed['Cd-D'], inplace=True)
dataprocessed['Fe-T'].fillna(dataprocessed['Fe-D'], inplace=True)
dataprocessed['Mn-T'].fillna(dataprocessed['Mn-D'], inplace=True)
dataprocessed['Mg-T'].fillna(dataprocessed['Mg-D'], inplace=True)
dataprocessed['Pb-T'].fillna(dataprocessed['Pb-D'], inplace=True)
dataprocessed['Ca-T'].fillna(dataprocessed['Ca-D'], inplace=True)
dataprocessed['K--T'].fillna(dataprocessed['K--D'], inplace=True)
dataprocessed['Hardness Total (T)'].fillna(dataprocessed['Hardness (Dissolved)'], inplace=True)
dataprocessed['N.Kjel:T'].fillna(dataprocessed['Nitrogen (Kjel.) Tot Diss'], inplace=True)
dataprocessed['P--T'].fillna(dataprocessed['Phosphorus Tot. Dissolved'], inplace=True)
dataprocessed['Alkalinity Total 4.5'].fillna(dataprocessed['Alkalinity pH 4.5/4.2'], inplace=True)
dataprocessed['Mg-T'] = dataprocessed['Mg-T'].astype(float)
dataprocessed['Ca-T'] = dataprocessed['Ca-T'].astype(float)
dataprocessed['Hardness Total (T)'].fillna(dataprocessed['Mg-T']*4.118 + dataprocessed['Ca-T']*2.497, inplace=True)
dataprocessed['pH'] = pd.to_numeric(dataprocessed['pH'], errors='coerce')

dataprocessed.head()
```
