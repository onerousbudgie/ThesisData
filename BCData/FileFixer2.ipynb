```python
import pandas as pd
climdata = pd.read_excel('Fused_data.xlsx')
stations = climdata['Station Name'].unique()
# len(stations)
```

```python
climdata.head()
```

```python
datawithObs = pd.read_excel('TimeWiseProcessed_data.xlsx')
datawithObs['LONGITUDE'] = - datawithObs['LONGITUDE']
datawithObs.head()
```

```python
datawithObs['pdtime'] = pd.to_datetime(datawithObs['DATE'])
```

```python
datawithObs['Month Year'] = datawithObs['pdtime'].dt.to_period('M')
datawithObs
```

```python
import numpy as np
dataprocessed = pd.DataFrame()
i = 1
wells = datawithObs['OBSWELL'].unique()
for well in wells:
    print(i, well)
    i += 1
    subset = datawithObs.loc[datawithObs['OBSWELL'] == well]
    times = subset['Month Year'].unique()
    for time in times:
        timeslice = subset.loc[subset['Month Year'] == time].copy()
        tempdf = pd.DataFrame()
        tempdf['LOCATION'] = [timeslice['LOCATION'].iloc[0]]
        tempdf['OBSWELL'] = [timeslice['OBSWELL'].iloc[0]]
        tempdf['LONGITUDE'] = [timeslice['LONGITUDE'].iloc[0]]
        tempdf['LATITUTE'] = [timeslice['LATITUTE'].iloc[0]]
        tempdf['DATE'] = [timeslice['Month Year'].iloc[0]]

        for param in timeslice.columns[6:-2]:
            tempdf[str(param)] = np.mean(timeslice[param])
        dataprocessed = pd.concat([dataprocessed, tempdf], ignore_index=True)
```

```python
dataprocessed.to_excel('AggregateTimewiseWells.xlsx')
```

```python
import math
blo = []
bla = []
bls = []
for index, row in dataprocessed.iterrows():
    if (index == 10):
        break
    print(index)

    latitude = row['LATITUTE']
    longitude = row['LONGITUDE']
    maxdist = np.inf
    bestlong = 0
    bestlat = 0
    beststation = ''
    for station in stations:
        subset = climdata.loc[climdata['Station Name'] == station, ['Longitude', 'Latitude']]
        stationlong = subset['Longitude'].iloc[0]
        stationlat = subset['Latitude'].iloc[0]
        d = math.dist([latitude, longitude], [stationlat, stationlong])
        if d < maxdist:
            maxdist = d
            bestlong = stationlong
            bestlat = stationlat
            beststation = station

    blo.append(bestlong)
    bla.append(bestlat)
    bls.append(beststation)


dataprocessed['Closest Station'] = bls
dataprocessed['Closest Longitude'] = blo
dataprocessed['Closest Latitude'] = bla
```

```python
dataprocessed.to_excel('MergedwithStation.xlsx')
```
